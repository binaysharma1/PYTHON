```python
x0,x1,x2=2.,3.,5,
bias,w1,w2=1,0.2,0.5
```

```python
x=[x0,x1,x2]


w=[bias,w1,w2]
```

```python
sum=0
for i in range(len(x)):
    sum+=x[i]*w[i]


print(sum)
```

```python
import numpy as np

x_vec,y_vec=np.array(x),np.array(w)

result=(x_vec.transpose()).dot(y_vec)
print(result)
```

```python
import time
size=10000
mat1_cpu=torch.rand(size,size)
mat2_cpu=torch.rand(size,size)

start_time=time.time()
torch.matmul(mat1_cpu,mat2_cpu)
cpu_time=time.time()-start_time

print("time taken by cpu :",cpu_time)
```

```python
mat1_gpu=mat1_cpu.to('cuda')
mat2_gpu=mat2_cpu.to('cuda')

start_time=time.time()
torch.matmul(mat1_gpu,mat2_gpu)
torch.cuda.synchronize() # Wait for GPU to finish computations
gpu_time=time.time()-start_time

print("time taken by gpu :",gpu_time)
```
